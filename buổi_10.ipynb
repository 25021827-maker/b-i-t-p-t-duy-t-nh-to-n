```python
#3
try:
    n=int(input())
    seen=set()
    while n !=1:
        if n in seen:
            print("NO")
            break
        seen.add(n)
        n=sum(int(i)**2 for i in str(n))
    else:
        print("YES")
except ValueError:
    print("phai nhap so nguyen")
```

```python
#10
a=list(map(int,input().split()))
try:
    seen=[]
    b=a[0]
    for i in range(1,b+1):
        if a[i] not in seen:
            seen.append(a[i])
    if sum(seen)==sum(a[1:]):
        print(sum(seen))
    else:
        raise ValueError
except (ValueError,IndexError):
    print("mang khong hop le")
```
